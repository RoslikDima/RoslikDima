```python
import pandas as pd

# Создаем серию с целыми числами и пропуском
s = pd.Series([1, 2, None, 4])
print(s.dtype)  # float64! Единица превратилась в 1.0
```

```python
# Теперь целые числа остаются целыми, а None — это нормальный null!
s_arrow = pd.Series([1, 2, None, 4], dtype="int64[pyarrow]")
print(s_arrow)
# 0       1
# 1       2
# 2    <NA>
# 3       4
# dtype: int64[pyarrow]
```

```python
cities = pd.Series(["Москва", "Казань", "СПб"], dtype="string[pyarrow]")
cities
```

```python
status = pd.Series(["new", "delivered", "new", "delivered"], dtype="category")
status
```
